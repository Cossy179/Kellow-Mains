# Decision Support Systems Notebook

## 1. Importing Libraries and Loading Data
In this section, we import the necessary libraries and load the dataset from a CSV file. A preview of the data is printed to confirm that the file has been loaded successfully.


```python
import pandas as pd
import numpy as np

# Load the dataset
file_path = "Kelloe mains Harvest_2023(2).csv"
df = pd.read_csv(file_path)

# Preview the dataset
print("Dataset preview:")
print(df.head())
```

## 2. Data Preprocessing
This section focuses on preparing the data for decision support analysis:
- A helper function is defined to convert columns to numeric values, properly handling non-numeric placeholders.
- Key columns related to farm performance (such as "Area Harvested", "Moisture", "Dry Yield", "Total Dry Yield", "Speed", "Total Fuel", "Fuel Efficiency", and "Crude Protein") are selected and converted.
- Rows with missing values in these important columns are dropped to ensure clean data for further analysis.


```python
# Helper function to convert columns to numeric values (handling non-numeric placeholders)
def to_numeric(col):
    return pd.to_numeric(col.replace('---', np.nan), errors='coerce')

# Define key columns for decision support analysis
columns_to_convert = [
    "Area Harvested", "Moisture", "Dry Yield", "Total Dry Yield",
    "Speed", "Total Fuel", "Fuel Efficiency", "Crude Protein"
]

# Convert selected columns to numeric
for col in columns_to_convert:
    df[col] = to_numeric(df[col])

# Drop rows with missing values in these key columns
df_clean = df.dropna(subset=columns_to_convert)
print("Cleaned data shape:", df_clean.shape)
```

## 3. Calculating Key Performance Indicators (KPIs)
Here, we compute the average values of critical metrics:
- Average Dry Yield
- Average Fuel Efficiency
- Average Crude Protein  
These KPIs serve as the foundation for generating actionable insights and guiding decisions.


```python
# Compute average values for key metrics that will drive decision-making
avg_dry_yield = df_clean["Dry Yield"].mean()
avg_fuel_efficiency = df_clean["Fuel Efficiency"].mean()
avg_crude_protein = df_clean["Crude Protein"].mean()

print("Average Dry Yield:", avg_dry_yield)
print("Average Fuel Efficiency:", avg_fuel_efficiency)
print("Average Crude Protein:", avg_crude_protein)
```

## 4. Rule-Based Decision Support System
This section implements a simple, rule-based decision support system by:
- Defining thresholds for key metrics based on domain knowledge.
- Comparing the computed KPIs against these thresholds.
- Generating recommendations based on whether the metrics meet the desired performance criteria.
Additional rules can be added or modified based on specific operational insights.


```python
# Define thresholds for key metrics (example values; adjust as needed based on domain expertise)
yield_threshold = 12.0         # Expected minimum for Dry Yield
fuel_eff_threshold = 8.0       # Expected minimum for Fuel Efficiency
protein_threshold = 25.0       # Expected minimum for Crude Protein

# Initialize a list to store recommendations
recommendations = []

# Provide recommendations based on average Dry Yield
if avg_dry_yield < yield_threshold:
    recommendations.append("Dry Yield is below the expected threshold. Consider reviewing field management practices and optimizing harvest timing.")
else:
    recommendations.append("Dry Yield is within the expected range.")

# Provide recommendations based on Fuel Efficiency
if avg_fuel_efficiency < fuel_eff_threshold:
    recommendations.append("Fuel Efficiency is lower than expected. Evaluate machine maintenance and adjust operational speed to improve fuel consumption.")
else:
    recommendations.append("Fuel Efficiency is satisfactory.")

# Provide recommendations based on Crude Protein levels
if avg_crude_protein < protein_threshold:
    recommendations.append("Crude Protein levels are below target, indicating potential quality issues. Consider revisiting fertilization and harvest practices.")
else:
    recommendations.append("Crude Protein levels are within acceptable limits.")

# Additional rules can be added here based on further analysis or domain-specific insights.
```

## 5. Displaying Decision Support Recommendations
Finally, the notebook prints out the recommendations generated from the rule-based analysis. These insights are intended to help guide farm management decisions such as reviewing field practices, optimizing harvest timing, or addressing quality concerns.







```python
print("\nDecision Support Recommendations:")
for rec in recommendations:
    print("- " + rec)
```
